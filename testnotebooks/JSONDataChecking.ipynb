```python
import json
import numpy as np

import sys
sys.path.append('../')

```

```python
json_file_path = '../nogit/json_data/predictions_full_dataset_100g.jsonl'
with open(json_file_path, 'r') as file:
    data = [json.loads(line) for line in file]  

print(f"Total entries in JSONL file: {len(data)}")
print(f"Sample entry: {data[0]}")
print(f"Keys in sample entry: {list(data[0].keys())}")
```

```python

BASE_KEY = 'base_circuit_tokens'
PQC_KEY = 'predicted_pqc_circuit_tokens'

valid_data = []
P = 3
Q = 10

for entry in data:
    base_tokens = entry.get(BASE_KEY, [])
    pqc_tokens = entry.get(PQC_KEY, [])

    pqc_params = [token[2][0] for token in pqc_tokens]
    K = len(pqc_params) // (P * Q)

    pqc_params_reshaped = np.array(pqc_params).reshape((K, Q, P))


    valid_data.append((base_tokens, pqc_params_reshaped))


valid_data[:2]  # Display first two entries for verification
```
