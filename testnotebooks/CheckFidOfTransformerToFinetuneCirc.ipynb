```python
import json
import os
import sys

sys.path.append('../')
```

```python
DATA_PATH = '../' + 'nogit/finetune_results/finetuned_1q_10g_results.jsonl'
```

```python
final_test_fid_list = []
finetune_pqc_fid_list = []
transformer_fid_list = []

with open(DATA_PATH, 'r') as f:
    for line in f:
        data = json.loads(line)
        final_test_fid_list.append(data['final_train_fidelity'])
        finetune_pqc_fid_list.append(data['test_fidelity_pqc_mean'])
        transformer_fid_list.append(data['test_fidelity_transformer_mean'])

print(f'Number of circuits: {len(finetune_pqc_fid_list)}')
```

```python
import matplotlib.pyplot as plt
import numpy as np  

max_val_fin = np.round(np.max(final_test_fid_list), 6)
min_val_fin = np.round(np.min(final_test_fid_list), 6)
if min_val_fin == max_val_fin:
    min_val_fin -= 0.000001  # To avoid zero range in histogram
print(f'Fidelity Value Range: {min_val_fin} to {max_val_fin}')
print(f'Mean: {np.mean(final_test_fid_list)}, Median: {np.median(final_test_fid_list)}')

hist_range = (min_val_fin, max_val_fin)
hist_bins = 50

# arr1, arr2, _ = plt.hist(good_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)

# Add percentage labels on top of each bar
plt.figure(figsize=(10, 5))
n, bins, patches = plt.hist(final_test_fid_list, bins=hist_bins, edgecolor='black', range=hist_range)
total = len(final_test_fid_list)
for i in range(len(patches)):
    height = patches[i].get_height()
    percentage = (height / total) * 100
    if percentage > 0:
        plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, f'{percentage:.2f}%', 
             ha='center', va='bottom', fontsize=8, rotation=45)
        
plt.xlabel('Fidelity Value')
plt.ylabel('Number of Circuits')
plt.title('Final Testing Fidelity Value Distribution of Fine-Tuned Circuits')
plt.show()
```
