```python
import os
import sys
import json
sys.path.append('../')
```

```python
DATA_DIR = '../' + 'nogit/quaternion/no_uncomp_data/lel_zz/rzrxrz/5q_50g_10blk_data/'
GOOD_FIDELITY_DIR = DATA_DIR + 'good_fidelity/'
POOR_FIDELITY_DIR = DATA_DIR + 'poor_fidelity/'

good_fidelity_val = []
for file in os.listdir(GOOD_FIDELITY_DIR):
    with open(file=GOOD_FIDELITY_DIR + file, mode='r') as f:
        data = json.load(f)
        good_fidelity_val.append(data['fidelity'])

poor_fidelity_val = []
for file in os.listdir(POOR_FIDELITY_DIR):
    with open(file=POOR_FIDELITY_DIR + file, mode='r') as f:
        data = json.load(f)
        poor_fidelity_val.append(data['fidelity'])

all_fidelity_val = good_fidelity_val + poor_fidelity_val
```

```python
import matplotlib.pyplot as plt
import numpy as np  

max_val_all = np.round(np.max(all_fidelity_val), 4)
min_val_all = np.round(np.min(all_fidelity_val), 4)
if min_val_all == max_val_all:
    min_val_all -= 0.0001  # To avoid zero range in histogram
print(f'Fidelity Value Range: {min_val_all} to {max_val_all}')
print(f'Total Circuits: {len(all_fidelity_val)}')
print(f'Average Fidelity Value: {np.round(np.mean(all_fidelity_val), 4)}')
print(f'Median Fidelity Value: {np.round(np.median(all_fidelity_val), 4)}')

hist_range = (min_val_all, max_val_all)
hist_bins = 10

# arr1, arr2, _ = plt.hist(good_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)

# Add percentage labels on top of each bar
plt.figure(figsize=(8, 4))
n, bins, patches = plt.hist(all_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)
total = len(all_fidelity_val)
for i in range(len(patches)):
    height = patches[i].get_height()
    percentage = (height / total) * 100
    if percentage > 0:
        plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, f'{percentage:.2f}%', 
             ha='center', va='bottom')
        
plt.xlabel('Fidelity Value')
plt.ylabel('Number of Circuits')
plt.title('All Fidelity Value Distribution of Fine-Tuned Circuits')
plt.show()
```
