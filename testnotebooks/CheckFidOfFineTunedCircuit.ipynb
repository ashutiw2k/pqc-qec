```python
import json
import os
import sys

sys.path.append('../')
```

```python
DATA_PATH = '../' + 'nogit/quaternion/no_uncomp_data/single_qubit/rzrxrz/1q_10g_5blk_data/'
GOOD_FID_PATH = DATA_PATH + 'good_fidelity/'
POOR_FID_PATH = DATA_PATH + 'poor_fidelity/'
```

```python
good_fidelity_val = []
for file in os.listdir(GOOD_FID_PATH):
    if file.endswith('.json'):
        with open(GOOD_FID_PATH + file, 'r') as f:
            data = json.load(f)
            good_fidelity_val.append(data['fidelity'])


poor_fidelity_val = []
for file in os.listdir(POOR_FID_PATH):
    if file.endswith('.json'):
        with open(POOR_FID_PATH + file, 'r') as f:
            data = json.load(f)
            poor_fidelity_val.append(data['fidelity'])

print("Good Fidelity Values: ", len(good_fidelity_val))
print("Poor Fidelity Values: ", len(poor_fidelity_val))

all_fidelity_val = good_fidelity_val + poor_fidelity_val
print("All Fidelity Values: ", len(all_fidelity_val))
```

```python
import matplotlib.pyplot as plt
import numpy as np  

max_val_all = np.round(np.max(all_fidelity_val), 3)
min_val_all = np.round(np.min(all_fidelity_val), 3)
if min_val_all == max_val_all:
    min_val_all -= 0.001  # To avoid zero range in histogram
print(f'Fidelity Value Range: {min_val_all} to {max_val_all}')

hist_range = (min_val_all, max_val_all)
hist_bins = 10

# arr1, arr2, _ = plt.hist(good_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)

# Add percentage labels on top of each bar
plt.figure(figsize=(8, 4))
n, bins, patches = plt.hist(all_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)
total = len(all_fidelity_val)
for i in range(len(patches)):
    height = patches[i].get_height()
    percentage = (height / total) * 100
    if percentage > 0:
        plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, f'{percentage:.2f}%', 
             ha='center', va='bottom')
        
plt.xlabel('Fidelity Value')
plt.ylabel('Number of Circuits')
plt.title('All Fidelity Value Distribution of Fine-Tuned Circuits')
plt.show()
```

```python
min_val_good = np.round(np.min(good_fidelity_val), 3)
max_val_good = np.round(np.max(good_fidelity_val), 3)
print(f'"Good" Fidelity Value Range: {min_val_good} to {max_val_good}')

hist_range = (min_val_good, max_val_good)
hist_bins = 10

# arr1, arr2, _ = plt.hist(good_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)

# Add percentage labels on top of each bar
plt.figure(figsize=(8, 4))
n, bins, patches = plt.hist(good_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)
total = len(good_fidelity_val)
for i in range(len(patches)):
    height = patches[i].get_height()
    percentage = (height / total) * 100
    if percentage > 0:
        plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, f'{percentage:.2f}%', 
             ha='center', va='bottom')
        
plt.xlabel('Fidelity Value')
plt.ylabel('Number of Circuits')
plt.title('Good Fidelity Value Distribution of Fine-Tuned Circuits')
plt.show()
```

```python
if len(poor_fidelity_val) > 0:
    min_val_poor = np.round(np.min(poor_fidelity_val), 3)
    max_val_poor = np.round(np.max(poor_fidelity_val), 3)
    print(f'"Poor" Fidelity Value Range: {min_val_poor} to {max_val_poor}')

    hist_range = (min_val_poor, max_val_poor)
    hist_bins = 10

    # arr1, arr2, _ = plt.hist(good_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)

    # Add percentage labels on top of each bar
    plt.figure(figsize=(8, 4))
    n, bins, patches = plt.hist(poor_fidelity_val, bins=hist_bins, edgecolor='black', range=hist_range)
    total = len(poor_fidelity_val)
    for i in range(len(patches)):
        height = patches[i].get_height()
        percentage = (height / total) * 100
        if percentage > 0:
            plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, f'{percentage:.2f}%', 
                 ha='center', va='bottom')
            
    plt.xlabel('Fidelity Value')
    plt.ylabel('Number of Circuits')
    plt.title('Poor Fidelity Value Distribution of Fine-Tuned Circuits')
    plt.show()
else:
    print("No poor fidelity values found to plot.")
```
