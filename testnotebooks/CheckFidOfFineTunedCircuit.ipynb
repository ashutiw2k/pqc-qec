```python
import json
import os
import sys

sys.path.append('../')
```

```python
DATA_PATH = '../' + 'nogit/quaternion/no_uncomp_data/single_qubit/rzrxrz/1q_5g_5blk_data/'
GOOD_FID_PATH = DATA_PATH + 'good_fidelity/'
```

```python
fids = []
for file in os.listdir(GOOD_FID_PATH):
    if file.endswith('.json'):
        with open(GOOD_FID_PATH + file, 'r') as f:
            data = json.load(f)
            fids.append(data['fidelity'])
```

```python
import matplotlib.pyplot as plt

arr1, arr2, _ = plt.hist(fids, bins=10, edgecolor='black', range=(0.99, 1.0))

for i in range(len(arr1)):
    print(f'Bin {arr2[i]}: {arr1[i]}')

# Add percentage labels on top of each bar
n, bins, patches = plt.hist(fids, bins=10, edgecolor='black', range=(0.99, 1.0))
total = len(fids)
for i in range(len(patches)):
    height = patches[i].get_height()
    percentage = (height / total) * 100
    plt.text(patches[i].get_x() + patches[i].get_width() / 2, height, f'{percentage:.1f}%', 
             ha='center', va='bottom')
```
